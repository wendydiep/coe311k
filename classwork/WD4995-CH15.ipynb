```python
import numpy as np
import matplotlib.pyplot as plt

step_sizes = np.arange(0.01, 1.05, 0.05)
x_start, x_end = 0, 2
initial_y = 1 

def dydx(x, y):
    return -2 * y

first_run_y = None
first_h = step_sizes[0]

plt.figure(figsize=(12, 8))

for idx, h in enumerate(step_sizes):
    num_steps = int((x_end - x_start) / h)
    x_vals = np.linspace(x_start, x_end, num_steps + 1)
    y_vals = np.zeros(num_steps + 1)
    y_vals[0] = initial_y

    y_exact = np.exp(-2 * x_vals)

    for i in range(num_steps):
        y_vals[i + 1] = y_vals[i] + h * dydx(x_vals[i], y_vals[i])

    if idx == 0:
        first_run_y = y_vals

    step_ratio = int(round(h / first_h))
    y_first_sampled = first_run_y[::step_ratio]
    min_len = min(len(y_vals), len(y_first_sampled))
    pointwise_errors = np.abs(y_vals[:min_len] - y_first_sampled[:min_len])

    total_error = np.abs(y_vals - y_exact).sum()

    print(f"\nStep size h = {h:.2f}")
    print(f"Total accumulated error: {total_error:.6f}")
    print("Current run vs first run errors:")
    for i in range(min_len):
        print(f"  x = {x_vals[i]:.2f} | Error = {pointwise_errors[i]:.6f}")

    plt.plot(x_vals, y_vals, label=f"h = {h:.2f}")

plt.plot(x_vals, np.exp(-2 * x_vals), 'k--', linewidth=2, label='Exact: exp(-2x)')

plt.xlabel('x')
plt.ylabel('y')
plt.title("Euler's Method Approximations for Different Step Sizes")
plt.legend()
plt.grid(True)
plt.show()
```
