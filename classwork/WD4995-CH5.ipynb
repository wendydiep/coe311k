# Gaussian Elimination

```python
import numpy as np
def my_gaussian_elimination(A, b):

    A = np.array(A, dtype=float) 
    b = np.array(b, dtype=float)

    n = len(b)

    for k in range(n - 1):

        pivot_index = np.argmax(abs(A[k:, k])) + k

        if A[pivot_index, k] == 0:
            raise ValueError("Matrix is singular.")

        A[[k, pivot_index]] = A[[pivot_index, k]]
        b[[k, pivot_index]] = b[[pivot_index, k]]

        for i in range(k + 1, n):
            multiplier = A[i, k] / A[k, k]
            A[i, k:] -= multiplier * A[k, k:]
            b[i] -= multiplier * b[k]

    x = np.zeros(n)

    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - np.dot(A[i, i+1:], x[i+1:])) / A[i, i]

    return x
```

```python
A = [[2, 3, -1], [4, -2, 2], [-2, 1, 2]]
b = [5, 6, -1]
solution = my_gaussian_elimination(A, b)
print(solution)
```

# LU Decomposition

```python
def my_lu_decomposition(A):
    A = A.astype(float)
    n = A.shape[0]

    L = np.eye(n)
    U = A.copy()

    for k in range(n - 1):

        if U[k, k] == 0:
            raise ValueError("Zero pivot encountered.")

        for i in range(k + 1, n):
            factor = U[i, k] / U[k, k]
            L[i, k] = factor
            U[i, k:] -= factor * U[k, k:]

    return L, U
```
