```python
import numpy as np
import matplotlib.pyplot as plt

spring_const = 1.0
step = 0.01
start_val = 0
end_val = 10
pos_init = 1.0
vel_init = 0.0

t_points = np.arange(start_val, end_val + step, step)
pos = np.zeros(len(t_points))
vel = np.zeros(len(t_points))

pos[0] = pos_init
vel[0] = vel_init

for idx in range(1, len(t_points)):
    prev_pos = pos[idx - 1]
    prev_vel = vel[idx - 1]

    pos[idx] = prev_pos + step * prev_vel
    vel[idx] = prev_vel + step * (-spring_const * prev_pos)

plt.figure(figsize=(10, 5))
plt.plot(t_points, pos, label='Position')
plt.plot(t_points, vel, linestyle='--', label='Velocity')
plt.xlabel('x')
plt.ylabel('Values')
plt.title("Euler Approximation of d²y/dx² = -k y")
plt.legend()
plt.grid()
plt.show(block=True)
```
