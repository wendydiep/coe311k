```python
import numpy as np

def leibniz_pi_estimate(n_terms):
    pi_approx = 0.0
    
    for k in range(n_terms):
        term = (-1)**k / (2*k + 1)
        pi_approx += term
    
    return 4 * pi_approx


max_terms = 5000

for n in range(1, max_terms + 1):
    current_value = leibniz_pi_estimate(n)
    
    if n % 100 == 0:
        relative_error = abs(current_value - np.pi) / np.pi
        print(f"After {n:4d} terms → Relative Error = {relative_error:.10f}")
```
