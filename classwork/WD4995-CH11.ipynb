# Cubic Spline & Polyfit

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

x_vals = np.array([0, 1, 2, 3, 4, 5], dtype=float)
y_vals = np.array([0, 1, 0, 1, 0, 1], dtype=float)

x_smooth = np.linspace(min(x_vals), max(x_vals), 200)

spline_model = CubicSpline(x_vals, y_vals, bc_type='natural')
y_spline = spline_model(x_smooth)

poly_degree = 3
poly_coeff = np.polyfit(x_vals, y_vals, poly_degree)

print("Polynomial coefficients:", poly_coeff)

poly_function = np.poly1d(poly_coeff)
y_poly = poly_function(x_smooth)
plt.figure()

plt.scatter(x_vals, y_vals, label="Original Data")
plt.plot(x_smooth, y_spline, label="Cubic Spline")
plt.plot(x_smooth, y_poly, label="Polynomial Fit (degree 3)")

plt.title("Polynomial Fit vs Cubic Spline")
plt.xlabel("x values")
plt.ylabel("y values")
plt.legend()

plt.show()
```
