```python
import numpy as np
import time
import matplotlib.pyplot as plt

def slow_multiply(mat1, mat2):
    n = mat1.shape[0]
    result = [[0 for _ in range(n)] for _ in range(n)]
    
    for row in range(n):
        for col in range(n):
            for idx in range(n):
                result[row][col] += mat1[row][idx] * mat2[idx][col]
    
    return result

current_size = 3

sizes_tested = []
slow_times = []
fast_times = []

for iteration in range(8):
    matrix_A = np.random.rand(current_size, current_size)
    matrix_B = np.random.rand(current_size, current_size)

    start_slow = time.perf_counter()
    slow_result = slow_multiply(matrix_A, matrix_B)
    end_slow = time.perf_counter()
    slow_times.append(end_slow - start_slow)

    start_fast = time.perf_counter()
    fast_result = np.dot(matrix_A, matrix_B)
    end_fast = time.perf_counter()
    fast_times.append(end_fast - start_fast)

    sizes_tested.append(current_size)

    current_size *= 2

plt.plot(sizes_tested, slow_times, marker='o', label='Custom (Naive)')
plt.plot(sizes_tested, fast_times, marker='o', label='NumPy Optimized')

plt.title("Matrix Multiplication Runtime Comparison")
plt.xlabel("Matrix Dimension (n x n)")
plt.ylabel("Execution Time (seconds)")
plt.yscale('log')
plt.legend()
plt.grid()

plt.show()
```
