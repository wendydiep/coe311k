```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}")
print(f"Intercept: {c}")

plt.scatter(X, Y, label="Data Points")

x_line = np.linspace(1, 5, 100)
y_line = m * x_line + c
plt.plot(x_line, y_line, label="Best Fit Line")

plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.title("Least Squares Best Fit")
plt.show()
```

```python
np.random.seed(0)
Y_random = np.random.rand(5) * 6

A = np.vstack([X, np.ones(len(X))]).T
m_rand, c_rand = np.linalg.lstsq(A, Y_random, rcond=None)[0]

print(f"New Slope: {m_rand}")
print(f"New Intercept: {c_rand}")

plt.scatter(X, Y_random, label="Random Data")
x_line = np.linspace(1, 5, 100)
plt.plot(x_line, m_rand * x_line + c_rand, label="Best Fit Line")

plt.legend()
plt.title("Least Squares with Randomized Data")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def my_line(x):
    return 3*x + 2

x = np.arange(0, 1.01, 0.01)

np.random.seed(1)
noise = np.random.normal(0, 0.2, len(x))

y = my_line(x) + noise

A = np.vstack([x, np.ones(len(x))]).T

m_fit, c_fit = np.linalg.lstsq(A, y, rcond=None)[0]

print(f"Original slope: 3")
print(f"Fitted slope: {m_fit}")
print(f"Original intercept: 2")
print(f"Fitted intercept: {c_fit}")

plt.scatter(x, y, s=10, label="Noisy Data")
plt.plot(x, my_line(x), label="Original Line")
plt.plot(x, m_fit*x + c_fit, label="Least Squares Fit")

plt.legend()
plt.title("Least Squares Recovers Original Equation")
plt.show()
```
