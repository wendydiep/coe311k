```python
import numpy as np
import matplotlib.pyplot as plt

k_val = 1.0

step_list = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

t0 = 0
t_end = 10
y_start = 1.0
v_start = 0.0

plt.figure(figsize=(12, 6))

for h in step_list:

    A = np.array([
        [1, h],
        [-h * k_val, 1]
    ])

    eig_vals = np.linalg.eigvals(A)
    magnitudes = np.abs(eig_vals)

    print(f"\nStep size h = {h}")
    print("Eigenvalue magnitudes:", magnitudes)

    if np.all(magnitudes < 1):
        print("Solution is Stable")
    else:
        print("Solution may not be Entirely Stable ;)")

    t_vals = np.arange(t0, t_end + h, h)
    y_vals = np.zeros(len(t_vals))
    v_vals = np.zeros(len(t_vals))

    y_vals[0] = y_start
    v_vals[0] = v_start

    for i in range(1, len(t_vals)):
        y_vals[i] = y_vals[i - 1] + h * v_vals[i - 1]
        v_vals[i] = v_vals[i - 1] - h * k_val * y_vals[i - 1]

    plt.plot(t_vals, y_vals, label=f"h={h}")

plt.xlabel("x")
plt.ylabel("y(x)")
plt.title("Euler Method Stability for Different Step Sizes")
plt.legend()
plt.grid()
plt.show(block=True)
```
