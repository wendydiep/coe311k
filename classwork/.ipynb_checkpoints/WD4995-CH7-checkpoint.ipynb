```python
import numpy as np
import time


def jacobi_method(A, b, start_guess, tolerance, max_steps):
    n = len(b)
    x = start_guess.astype(float).copy()

    for step in range(max_steps):
        x_new = np.zeros(n)

        for i in range(n):
            total = 0
            for j in range(n):
                if j != i:
                    total += A[i][j] * x[j]

            x_new[i] = (b[i] - total) / A[i][i]

        if np.linalg.norm(x_new - x, ord=np.inf) < tolerance:
            return x_new, step

        x = x_new

    return x, max_steps


def gauss_seidel_method(A, b, start_guess, tolerance, max_steps):
    n = len(b)
    x = start_guess.astype(float).copy()

    for step in range(max_steps):
        x_new = x.copy()

        for i in range(n):

            sum_before = 0
            for j in range(i):
                sum_before += A[i][j] * x_new[j]

            sum_after = 0
            for j in range(i + 1, n):
                sum_after += A[i][j] * x[j]

            x_new[i] = (b[i] - sum_before - sum_after) / A[i][i]

        if np.linalg.norm(x_new - x, ord=np.inf) < tolerance:
            return x_new, step

        x = x_new

    return x, max_steps
```

```python
A = np.array([
    [0.582745, 0.48,     0.10,     0.0,      0.0],
    [0.48,     1.044129, 0.46,     0.10,     0.0],
    [0.10,     0.46,     1.10431,  0.44,     0.10],
    [0.0,      0.10,     0.44,     0.963889, 0.42],
    [0.0,      0.0,      0.10,     0.42,     0.522565]
], dtype=float)

b = np.array([
    1.162745,
    2.084129,
    2.20431,
    1.923889,
    1.042565
], dtype=float)

initial_guess = np.zeros(len(b))
tolerance = 1e-6
max_steps = 100

start_time = time.perf_counter()

jacobi_solution, jacobi_iters = jacobi_method(
    A, b, initial_guess, tolerance, max_steps
)

end_time = time.perf_counter()

print("Jacobi Solution:", jacobi_solution)
print("Jacobi Iterations:", jacobi_iters)
print(f"Jacobi Runtime: {(end_time - start_time):.6e} seconds")

start_time = time.perf_counter()

gs_solution, gs_iters = gauss_seidel_method(
    A, b, initial_guess, tolerance, max_steps
)

end_time = time.perf_counter()

print("Gauss-Seidel Solution:", gs_solution)
print("Gauss-Seidel Iterations:", gs_iters)
print(f"Gauss-Seidel Runtime: {(end_time - start_time):.6e} seconds")
```
