```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.subplot(2, 2, 1) # (row, column, active area)
plt.plot(x, np.sin(x), 'r')
plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x), 'g')
plt.subplot(2,2,3)
plt.plot(x,np.tan(x), 'k')
plt.subplot(2,2,4)
plt.plot(x, np.arcsin(x))
plt.show()
```

```python
x = np.linspace(0, 10, 50)
y = np.sin(x)
random_data = np.random.randn(500)

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.scatter(x, y)
plt.title("Scatter Plot")

plt.subplot(2, 2, 2)
plt.hist(random_data, bins=20)
plt.title("Histogram")

plt.subplot(2, 2, 3)
sizes = [25, 30, 20, 25]
labels = ['A', 'B', 'C', 'D']
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart")

plt.subplot(2, 2, 4)
plt.plot(x, np.cos(x))
plt.title("Line Plot")

plt.suptitle("Challenge 1 - 4 Plot Types", fontsize=14)
plt.show()

```
