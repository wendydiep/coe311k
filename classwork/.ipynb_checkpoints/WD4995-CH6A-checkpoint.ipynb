# Jacobi

```python
import numpy as np
import time


def jacobi_solver(A, b, initial_guess, tolerance, max_iter):
    n = len(b)
    x_current = initial_guess.copy()

    for iteration in range(max_iter):

        x_next = np.zeros_like(x_current)

        for row in range(n):

            sigma = 0
            for col in range(n):
                if col != row:
                    sigma += A[row, col] * x_current[col]

            x_next[row] = (b[row] - sigma) / A[row, row]

        change = np.linalg.norm(x_next - x_current, ord=np.inf)
        if change < tolerance:
            return x_next, iteration

        x_current = x_next

    return x_current, max_iter

A = np.array([
    [10, -2, 1, 0, 0],
    [-1, 11, -1, 3, 0],
    [2, -1, 10, -1, 0],
    [0, 3, -1, 8, -2],
    [0, 0, 0, -2, 9]
])

b = np.array([14, 30, 26, 25, 37])

initial_guess = np.zeros_like(b)
tolerance = 1e-6
max_iter = 100

start_time = time.perf_counter()

solution, iters = jacobi_solver(A, b, initial_guess, tolerance, max_iter)

end_time = time.perf_counter()
runtime = end_time - start_time


print("Computed Solution:", solution)
print("Iterations Needed:", iters)
print(f"Runtime: {runtime:.6e} seconds")
```

# Gauss-Seidel

```python
import numpy as np
import time


def gauss_seidel_method(A, b, start_guess, tolerance, max_steps):
    n = len(b)
    x = start_guess.copy()

    for step in range(max_steps):

        x_updated = x.copy()

        for i in range(n):

            sum_before = 0
            for j in range(i):
                sum_before += A[i][j] * x_updated[j]

            sum_after = 0
            for j in range(i + 1, n):
                sum_after += A[i][j] * x[j]

            x_updated[i] = (b[i] - sum_before - sum_after) / A[i][i]

        error = np.linalg.norm(x_updated - x, ord=np.inf)
        if error < tolerance:
            return x_updated, step

        x = x_updated

    return x, max_steps

A = np.array([
    [10, -2, 1, 0, 0],
    [-1, 11, -1, 3, 0],
    [2, -1, 10, -1, 0],
    [0, 3, -1, 8, -2],
    [0, 0, 0, -2, 9]
])

b = np.array([14, 30, 26, 25, 37])

start_guess = np.zeros_like(b)
tolerance = 1e-6
max_steps = 100

start_time = time.perf_counter()

solution, steps = gauss_seidel_method(A, b, start_guess, tolerance, max_steps)

end_time = time.perf_counter()
time_taken = end_time - start_time


print("Solution:", solution)
print("Iterations:", steps)
print(f"Operation Time: {time_taken:.6e} seconds")
```

# The Gauss-Seidel method reaches the solution faster than the Jacobi method. It takes fewer iterations to converge, and the error in the solution decreases more quickly compared to Jacobi.
