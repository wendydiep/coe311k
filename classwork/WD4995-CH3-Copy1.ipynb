```python
import random
import numpy as np
import matplotlib.pyplot as plt

def monte_carlo_pi(num_samples):
    inside_count = 0

    for _ in range(num_samples):
        x_coord = random.random()
        y_coord = random.random()

        if x_coord**2 + y_coord**2 <= 1:
            inside_count += 1

    pi_estimate = 4 * inside_count / num_samples
    return pi_estimate

total_runs = 100
samples_per_run = 1000

abs_error_vals = []
rel_error_vals = []
run_index = []

for run in range(1, total_runs + 1):
    pi_calc = monte_carlo_pi(samples_per_run)
    true_pi = np.pi

    absolute_error = abs(true_pi - pi_calc)
    relative_error = absolute_error / true_pi

    run_index.append(run)
    abs_error_vals.append(absolute_error)
    rel_error_vals.append(relative_error)


plt.figure(figsize=(12, 6))
plt.plot(run_index, abs_error_vals, label="Absolute Error", marker='o')
plt.plot(run_index, rel_error_vals, label="Relative Error", marker='s')

plt.xlabel("Run Number")
plt.ylabel("Error Magnitude")
plt.title("Monte Carlo Pi Approximation: Absolute vs Relative Error")
plt.legend()
plt.grid(True)
plt.show()
```
