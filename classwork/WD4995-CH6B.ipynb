# Jacobi

```python
import numpy as np
import time


def jacobi_solver(A, b, guess, tol, max_steps):

    n = len(b)
    x_old = guess.astype(float).copy()  
    for step in range(max_steps):

        x_new = np.zeros(n)

        for i in range(n):

            total = 0
            for j in range(n):
                if j != i:
                    total += A[i][j] * x_old[j]

            x_new[i] = (b[i] - total) / A[i][i]

        if np.linalg.norm(x_new - x_old, ord=np.inf) < tol:
            return x_new, step

        x_old = x_new

    return x_old, max_steps

A = np.array([
    [1, 2, 3, 0, 0],
    [2, 1, 2, 3, 0],
    [3, 2, 1, 2, 3],
    [0, 3, 2, 1, 2],
    [0, 0, 3, 2, 1]
], dtype=float)

b = np.array([14, 22, 33, 26, 22], dtype=float)

initial_guess = np.zeros(len(b))

tol = 1e-6
max_steps = 100


start_time = time.perf_counter()

solution, iterations = jacobi_solver(A, b, initial_guess, tol, max_steps)

end_time = time.perf_counter()

print("Solution:", solution)
print("Iterations:", iterations)
print("Operation Time:", end_time - start_time)
```

# Gauss-Seidel

```python
import numpy as np
import time


def gauss_seidel_solver(A, b, start_guess, tol, max_steps):

    n = len(b)

    x = start_guess.astype(float).copy()

    for step in range(max_steps):

        x_next = x.copy()

        for i in range(n):

            part1 = 0
            for j in range(i):
                part1 += A[i][j] * x_next[j]

            part2 = 0
            for j in range(i + 1, n):
                part2 += A[i][j] * x[j] 

            x_next[i] = (b[i] - part1 - part2) / A[i][i]

        if np.linalg.norm(x_next - x, ord=np.inf) < tol:
            return x_next, step

        x = x_next

    return x, max_steps

A = np.array([
    [1, 2, 3, 0, 0],
    [2, 1, 2, 3, 0],
    [3, 2, 1, 2, 3],
    [0, 3, 2, 1, 2],
    [0, 0, 3, 2, 1]
], dtype=float)

b = np.array([14, 22, 33, 26, 22], dtype=float)

start_guess = np.zeros(len(b))

tol = 1e-6
max_steps = 100


start_time = time.perf_counter()

solution, iterations = gauss_seidel_solver(A, b, start_guess, tol, max_steps)

end_time = time.perf_counter()


print("Solution:", solution)
print("Iterations:", iterations)
print(f"Operation Time: {end_time - start_time:.6e} seconds")
```

# The Gauss-Seidel method reaches the solution more quickly than the Jacobi method. It needs fewer iterations to get there, and the values approach the final answer faster compared to Jacobi.
