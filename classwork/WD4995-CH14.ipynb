```python
import numpy as np
import matplotlib.pyplot as plt

x_start = 0
x_end = 2 * np.pi
num_points = 30

x_vals = np.linspace(x_start, x_end, num_points)
step = (x_end - x_start) / (num_points - 1)

true_derivative = np.cos(x_vals)

central_diff = (np.sin(x_vals + step) - np.sin(x_vals - step)) / (2 * step)
forward_diff = (np.sin(x_vals + step) - np.sin(x_vals)) / step
backward_diff = (np.sin(x_vals) - np.sin(x_vals - step)) / step

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].plot(x_vals, true_derivative, color='black', label='cos(x)', linewidth=2)
axes[0, 0].plot(x_vals, forward_diff, color='red', linestyle='-', label='Forward Diff')
axes[0, 0].set_title('Forward Difference Approximation')
axes[0, 0].legend()
axes[0, 0].grid(True)

axes[0, 1].plot(x_vals, true_derivative, color='black', label='cos(x)', linewidth=2)
axes[0, 1].plot(x_vals, backward_diff, color='blue', linestyle='--', label='Backward Diff')
axes[0, 1].set_title('Backward Difference Approximation')
axes[0, 1].legend()
axes[0, 1].grid(True)

axes[1, 0].plot(x_vals, true_derivative, color='black', label='cos(x)', linewidth=2)
axes[1, 0].plot(x_vals, central_diff, color='green', linestyle='-.', label='Central Diff')
axes[1, 0].set_title('Central Difference Approximation')
axes[1, 0].legend()
axes[1, 0].grid(True)

axes[1, 1].plot(x_vals, true_derivative, color='black', linewidth=2, label='cos(x)')
axes[1, 1].plot(x_vals, forward_diff, color='red', linestyle='-', label='Forward Diff')
axes[1, 1].plot(x_vals, backward_diff, color='blue', linestyle='--', label='Backward Diff')
axes[1, 1].plot(x_vals, central_diff, color='green', linestyle='-.', label='Central Diff')
axes[1, 1].set_title('Comparison of Numerical Methods')
axes[1, 1].legend()
axes[1, 1].grid(True)

plt.tight_layout()
plt.show(block=True)
```
