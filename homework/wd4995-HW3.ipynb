# Taylor Series of Sin(X)

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def taylor_sin(x, terms):
    total = 0
    
    for n in range(terms):
        num = (-1)**n * x**(2*n + 1)
        den = math.factorial(2*n + 1)
        total += num / den
        
    return total
```

# Approximate sin⁡(x) at specific values of x.

```python
x_values = np.arange(0, 10.1, 0.1)

sin_1term = taylor_sin(x_values, 1)
sin_2term = taylor_sin(x_values, 2)
sin_3term = taylor_sin(x_values, 3)
sin_4term = taylor_sin(x_values, 4)

sin_exact = np.sin(x_values)
```

# Calculate absolute value

```python
error_1 = np.abs(sin_1term - sin_exact)
error_2 = np.abs(sin_2term - sin_exact)
error_3 = np.abs(sin_3term - sin_exact)
error_4 = np.abs(sin_4term - sin_exact)
```

# Plot the truncation error

```python
plt.plot(x_values, error_1, label="1 Term")
plt.plot(x_values, error_2, label="2 Terms")
plt.plot(x_values, error_3, label="3 Terms")
plt.plot(x_values, error_4, label="4 Terms")

plt.xlabel("x values")
plt.ylabel("Absolute Error")
plt.title("Truncation Error of Taylor Series for sin(x)")
plt.legend()
plt.show()
```

# As more terms are included in the Taylor series expansion, the truncation error generally decreases because the polynomial approximation becomes a better representation of the true sin(x) function. Each added term improves the curve’s accuracy, especially near x = 0. However, the error does not decrease uniformly for all values of x. For smaller values of x, the approximation converges quickly and the error becomes very small even with just a few terms. For larger values of x (closer to 10), the error increases significantly, especially when using only 1 or 2 terms. This happens because higher powers of x grow rapidly, making the truncated polynomial less accurate far from the expansion point (which is centered at 0).
