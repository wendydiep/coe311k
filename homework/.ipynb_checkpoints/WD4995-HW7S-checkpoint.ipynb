```python
import numpy as np
import matplotlib.pyplot as plt

step_sizes = [0.01, 0.05, 0.15, 0.20, 0.25, 0.50, 0.75, 1.00]

x_vals = np.linspace(0, 10*np.pi, 1000)
lambda_vals = -2 * np.cos(x_vals)

plt.figure(figsize=(12,5))

for h in step_sizes:
    amplification = np.abs(1 + h * lambda_vals)
    plt.plot(x_vals/np.pi, amplification, label=f"h = {h}")

plt.axhline(y=1.0, linestyle='--', label="stability boundary")

plt.title("Euler Amplification Factor")
plt.xlabel("x / π")
plt.ylabel("|1 + hλ|")
plt.legend()
plt.grid()

plt.savefig("plot1_amplification.png") 
plt.show(block=True) 

stable_ratio = []

for h in step_sizes:
    amplification = np.abs(1 + h * lambda_vals)
    stable_ratio.append(np.mean(amplification <= 1))

plt.figure(figsize=(8,5))
plt.bar([str(h) for h in step_sizes], stable_ratio)

plt.axhline(y=1.0, linestyle='--', label="stable")

plt.title("Stable Fraction Over Interval")
plt.xlabel("h")
plt.ylabel("fraction stable")
plt.legend()
plt.grid()

plt.savefig("plot2_stability_fraction.png") 
plt.show(block=True)
```
